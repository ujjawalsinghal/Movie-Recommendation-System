```python
def convert_genres(text):
    genre_list = ast.literal_eval(text)

    name = []
    for genre in genre_list:
        name.append(genre["name"])

    return " ".join(name)
```

```python
convert_genres(movies.loc[0, "genres"])
```

```python
movies["genres"] = movies["genres"].apply(convert_genres)
```

```python
movies[["title", "genres"]].head()
```

```python
def convert_names(text):
    items = ast.literal_eval(text)

    names = []

    for item in items:
        names.append(item["name"])
    return " ".join(names)
```

```python
movies["keywords"] = movies["keywords"].apply(convert_names)
```

```python
movies[["title", "keywords"]].head()
```

```python
movies[["title", "genres", "keywords"]].head()
```

```python
def convert_cast(text):
    cast_list = ast.literal_eval(text)

    names = []

    for cast in cast_list[:3]:  # Limit to the first 3 cast members
        names.append(cast["name"])
    return " ".join(names)
```

```python
movies["cast"] = credits["cast"].apply(convert_cast)
```

```python
movies[["title", "cast"]].head()
```

```python
def get_director(text):
    crew_list = ast.literal_eval(text)

    for crew in crew_list:
        if crew["job"] == "Director":
            return crew["name"]
    return ""
```

```python
movies["director"] = credits["crew"].apply(get_director)
```

```python
movies = movies[[ "title", "genres", "keywords", "overview", "cast", "director"]]
```

```python
movies["tags"] = (
    movies["genres"] + " " + 
    movies["keywords"] + " " + 
    movies["overview"] + " " + 
    movies["cast"] + " " + 
    movies["director"]
)
```

```python
vectorizer = TfidfVectorizer(max_features=5000)
```

```python
movie_indices = pd.Series(movies.index, index=movies["title"]).drop_duplicates()
```

```python
movie_indices["Avatar"]
```

```python
def recommend(movie):
    movie = movie.lower()

    title_map = {
        title.lower(): title
        for title in movies["title"]
    }

    if movie not in title_map:
        print("Movie not found in the dataset.")
        return []

    actual_title = title_map[movie]
    index = movie_indices[actual_title]

    similar_movies = sorted(
        list(enumerate(similarity[index])),
        key=lambda x: x[1],
        reverse=True
    )

    recommendations = []

    for i in similar_movies[1:6]:
        recommendations.append({
            "title": movies.iloc[i[0]]["title"],
            "score": float(round(i[1], 3))
        })

    return recommendations
```
